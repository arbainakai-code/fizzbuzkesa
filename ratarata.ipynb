```
mahasiswa = {
    "asep": [50, 70, 40, 80],
    "budi": [78, 78, 80, 65],
    "cecep": [57, 88, 67, 69]
}

for nama, nilai in mahasiswa.items():
    ratarata = sum(nilai) / 4
    nilaiterkecil = min(nilai)

    print("rata-rata nilai", nama, "adalah:", ratarata)
    print("nilai terkecil", nama, "adalah:", nilaiterkecil)
    print()

```
